# Challenge 6 — Group 3: Public Health & Epidemiology
## AutoEncoders, Variational AutoEncoders, and Isolation Forest

This notebook reproduces the Challenge 6 workflow for Group 3 using the same preprocessed BRFSS feature matrix from Challenge 5.

Workflow:
1. Load Challenge 5 processed features and K-Means labels.
2. Train a standard AutoEncoder (AE).
3. Train a Variational AutoEncoder (VAE).
4. Fit an Isolation Forest baseline.
5. Save anomaly scores, latent-space summaries, and comparison figures.

The generated outputs are written to `results/challenge6`, `figures/challenge6`, and `weights/challenge6`.

## 1. Setup
Add the repository `src/` folder to `sys.path` so the pipeline module can be imported from the notebook.

```python
from pathlib import Path
import json
import sys

import pandas as pd

repo_root = Path.cwd().resolve()
while repo_root != repo_root.parent and not (repo_root / 'src' / 'pipeline.py').exists():
    repo_root = repo_root.parent

src_dir = repo_root / 'src'
if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

from pipeline import Challenge6Config, run

print(f'Repository root: {repo_root}')
print(f'Source directory: {src_dir}')
```

## 2. Run the Challenge 6 pipeline
This cell executes the full pipeline using the Challenge 5 artefacts stored in `results/challenge5`.

If you want to inspect the notebook without rerunning training, set `RUN_PIPELINE = False`.

```python
RUN_PIPELINE = True

config = Challenge6Config(
    ch5_results_dir=repo_root / 'results' / 'challenge5',
    output_dir=repo_root / 'results' / 'challenge6',
    figures_dir=repo_root / 'figures' / 'challenge6',
    weights_dir=repo_root / 'weights' / 'challenge6',
)

if RUN_PIPELINE:
    run(config)
else:
    print('Pipeline run skipped.')
```

## 3. Inspect generated outputs
Load the summary tables and top-anomaly files created by the pipeline.

```python
output_dir = repo_root / 'results' / 'challenge6'
summary_path = output_dir / 'summary_table.csv'
metrics_path = output_dir / 'metrics_by_seed.csv'
best_config_path = output_dir / 'best_config.json'

if summary_path.exists():
    display(pd.read_csv(summary_path))
else:
    print(f'Missing: {summary_path}')

if metrics_path.exists():
    display(pd.read_csv(metrics_path).head())
else:
    print(f'Missing: {metrics_path}')

if best_config_path.exists():
    with open(best_config_path, 'r', encoding='utf-8') as f:
        print(json.dumps(json.load(f), indent=2))
else:
    print(f'Missing: {best_config_path}')
```
